```python
#ejercicio 2:

nombre=["sofia","karen","lucia","julian","mayra"]
filtro=[]
for n in nombre:
  if n[0]=="m": filtro.append(n)
filtro
```

```python
f2=[n for n in nombre if n[0]=="m"]
f2
```

```python
#imprimir nombres con longitud>5
f3=[n for n in nombre if len(n)>=6]
f3
```

```python
# imprimir los nombres de los estudiantes que pasaron
nota=[2.5,3.2,0.1,4.9,3]
f4=[z for z in nota if z>=3]
f4
```
